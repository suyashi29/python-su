### Discussion
NumPy generates random values using mathematical rules. This is not AI.

```python
import numpy as np
import pandas as pd

np.random.seed(42)
customers=pd.DataFrame({
"Customer_ID":range(1,11),
"Age":np.random.randint(20,60,10),
"Salary":np.random.randint(30000,150000,10),
"Purchase_Amount":np.random.randint(500,5000,10)})
customers
```

```python
# Uncomment if Faker is not installed
# !pip install faker
from faker import Faker
import pandas as pd

fake=Faker()
records=[]
for i in range(11):
    records.append({
        "Name":fake.name(),
        "Email":fake.email(),
        "City":fake.city(),
        "Company":fake.company()
    })
pd.DataFrame(records)
```

```python
from faker import Faker
import numpy as np
import pandas as pd

fake=Faker()
students=[]
for i in range(20):
    students.append({
        "Student_Name":fake.name(),
        "Age":np.random.randint(18,25),
        "Course":np.random.choice(["Python","AI","Data Science","GAN"])
    })
pd.DataFrame(students)
```

```python
import numpy as np
letters=list("ABCDEFGHIJKLMNOPQRSTUVWXYZ")
generated_word="".join(np.random.choice(letters,size=5))
print("Generated Text:",generated_word)
```

```python
import numpy as np
characters=list("GANMODEL")
for i in range(5):
    word="".join(np.random.choice(characters,size=6))
    print(word)
```

```python
import matplotlib.pyplot as plt
import numpy as np
ages=np.random.randint(18,60,500)
plt.hist(ages,bins=10)
plt.title("Generated Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

## NumPy/Faker vs GANs

- NumPy/Faker follow predefined rules.
- GANs learn patterns from real data.
- NumPy/Faker do not train.
- GANs require training using Generator and Discriminator networks.
